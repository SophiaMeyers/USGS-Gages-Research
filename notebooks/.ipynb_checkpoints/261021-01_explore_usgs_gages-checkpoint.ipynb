### Downloading the active_flow Gages_sumary_wy.rd

```python
from pathlib import Path
import requests

url = "https://raw.githubusercontent.com/DOI-USGS/gages-through-the-ages/main/gage_timeseries_gif/active_flow_gages_summary_wy.rds"

output_path = Path("../data/active_flow_gages_summary_wy.rds")
output_path.parent.mkdir(parents=True, exist_ok=True)

response = requests.get(url, timeout=60)
response.raise_for_status()

output_path.write_bytes(response.content)

print("Downloaded successfully")
print("Saved to:", output_path.resolve())
print("File size:", f"{output_path.stat().st_size:,}", "bytes")
    
```
